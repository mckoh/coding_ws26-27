# Übung 01: Megatutorial

* **Hallo Welt & Steckbrief:** Wir legen in VS Code unsere erste Python-Datei an, speichern unseren Namen, unser Alter und unsere Körpergröße in passenden Variablen (`str`, `int`, `float`) und geben diese über `print()` gemeinsam auf der Konsole aus.
* **Interaktiver Begrüßungs-Generator:** Wir erweitern unser Skript um die `input()`Funktion, fragen den Namen und den Wohnort der Nutzenden ab und geben mithilfe eines **f-Strings** eine personalisierte Willkommensnachricht aus.
* **Der einfache Trinkgeldrechner:** Wir fragen über `input()` einen Rechnungsbetrag ab, wandeln die Eingabe mit `float()` in eine Gleitkommazahl um, berechnen 10 % Trinkgeld sowie den Gesamtbetrag und geben das Ergebnis gerundet aus.
* **Text-Detektiv (String-Indexierung & Slicing):** Wir speichern einen längeren Satz in einer Variable und extrahieren gemeinsam über eckige Klammern `[]` den ersten Buchstaben, den letzten Buchstaben sowie ein bestimmtes Wort aus der Mitte des Satzes.
* **E-Mail-Normalisierer:** Wir nehmen eine unsauber formatierte E-Mail-Adresse (z. B. `" Max.Mustermann@MCI.edu "`) und bereinigen sie schrittweise mit den String-Methoden `.strip()` und `.lower()`, bevor wir mit `.replace()` die Domain anpassen.

```python
# a) Hallo Welt & Steckbrief

name = "Michael"
age = 20
size = 1.65

print(f"Name: {name}, Alter: {age}, Körpergröße: {size}")
```

```python
# b) Interaktiver Begrüßungs-Generator

input_name = input("Bitte gib deinen Namen ein.")
input_location = input("Wo wohnst du?")

print(f"Name: {input_name}, Wohnort: {input_location}")
```

```python
# c) Der einfache Trinkgeldrechner

amount = input("Gib den Rechnungsbetrag ein (z. B. 35.90).")

amount = float(amount)
amount = amount * 1.1
amount = round(amount, 2)

# Alternative Form
# amount = round(float(amount) * 1.1, 2)

print(f"Zu bezahlender Betrag inkl. Trinkgeld: {amount}.")
```

```python
# d) Text-Detektiv (String-Indexierung & Slicing)

sentence = "Heute ist es sehr kalt."
print(sentence[0]) # Erster Buchstabe
print(sentence[6:9]) # Wort in der Satzmitte
# print(sentence[-10:-1]) # Satzteil von hinten aus referenziert
# print(sentence[-10:]) # Satzteil bis zum Satzende
print(sentence[-1]) # Letzter Buchstabe
```

```python
# e) E-Mail-Normalisierung

email = "    Michael_Kohlegger@MCI_edu  "

email = email.strip()
email = email.lower()
email = email.replace("_", ".")

# email = email.strip().lower().replace("_", ".")

print(f"E-Mail-Adresse ist {email}.")
```
